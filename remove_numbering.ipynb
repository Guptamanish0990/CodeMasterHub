```python
# Remove Leading Numbering from problems.js Titles
import re
from pathlib import Path

problem_files = list(Path('src/data').rglob('problems.js'))
print(f'Found {len(problem_files)} problems.js files')

pattern = re.compile(r'^(\s*(?:title|question)\s*:\s*["\"])\d+\.\s*', re.MULTILINE)
for file in problem_files:
    text = file.read_text(encoding='utf-8')
    new_text, count = pattern.subn(r'\1', text)
    if count:
        file.write_text(new_text, encoding='utf-8')
        print(f'Updated {file}: {count} replacements')
```
